```python
# ---- 先前在 Mission 2 中定義的變數（若未執行過該 cell 或 kernel 已重開，請先執行這兩行）----
import numpy as np
import pandas as pd

student_count = 60                      # 學生總人數
rng = np.random.default_rng(42)         # 亂數產生器（seed 請依你 Mission 2 中設定的數值）
# --------------------------------------------------------------------------

n = student_count

# 學號：從 1 到 n
student_id = np.arange(1, n + 1)

# 科系：從三個選項中隨機選擇
departments_list = ["資訊管理", "企業管理", "金融管理"]
department = rng.choice(departments_list, size=n)

# 睡眠時數：平均 6.7，標準差 1.1，保留一位小數
sleep_hours = rng.normal(loc=6.7, scale=1.1, size=n)
sleep_hours = np.round(np.clip(sleep_hours, 0, 24), 1)

# 螢幕使用時數：平均 5.0，標準差 1.5，限制在 1~10 小時，保留一位小數
screen_time_hours = rng.normal(loc=5.0, scale=1.5, size=n)
screen_time_hours = np.round(np.clip(screen_time_hours, 1, 10), 1)

# 讀書時數：平均 2.5，標準差 1.0，限制在 0.5~6 小時，保留一位小數
study_hours = rng.normal(loc=2.5, scale=1.0, size=n)
study_hours = np.round(np.clip(study_hours, 0.5, 6), 1)

# 出席率：平均 88，標準差 8，限制在 60~100，四捨五入成整數
attendance_rate = rng.normal(loc=88, scale=8, size=n)
attendance_rate = np.round(np.clip(attendance_rate, 60, 100)).astype(int)

df = pd.DataFrame({
    "student_id": student_id,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

# 顯示前五筆資料
df.head()
```

```python
# #w6Hm1
# 1️⃣ 基礎分數
base = 50
# 2️⃣ 各項影響（先算出來，方便閱讀）
study_effect   = 3 * df["study_hours"]
sleep_effect   = 2 * df["sleep_hours"]
screen_effect  = 1.2 * df["screen_time_hours"]
attendance_effect = 0.15 * df["attendance_rate"]
# 3️⃣ 隨機誤差：平均 0、標準差 5
random_error = rng.normal(loc=0.0, scale=5.0, size=n)
# 4️⃣ 合計未處理的分數
raw_quiz = base + study_effect + sleep_effect - screen_effect + attendance_effect + random_error
# 5️⃣ 限制在 0~100 並四捨五入成整數
df["quiz_score"] = np.round(np.clip(raw_quiz, 0, 100)).astype(int)

df.index = np.arange(1, len(df) + 1)
# 顯示全部 60 份資料（把 .head() 拿掉即可）
df[
    [
        "student_id",
        "study_hours",
        "sleep_hours",
        "screen_time_hours",
        "attendance_rate",
        "quiz_score"
    ]
]
```
